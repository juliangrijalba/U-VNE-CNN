# Original Mel generation

- We use the same previously labeled audio segments to convert them into Mel spectrograms. In this way, we will have Mels with their respective (multi) labels (TM, MH, VO, MU, EM, SO, BI, GE, SI)

```python
import os
import librosa
import numpy as np
from PIL import Image
import matplotlib.cm as cm
```

# 1. Basic setup

```python
# INPUT
P1_ORG_AUDIO_DIR = r"E:\Clon TSI\P1\recortados_3min\normalizados\segmentos_2s_todos"
P2_ORG_AUDIO_DIR = r"E:\Clon TSI\P2\recortados_3min\normalizados\segmentos_2s_todos"

# OUTPUT
DPY_MELS_DIR = r"C:\Users\Acer\Desktop\TSI\U_VNE_CNN\5_cnn_deployment_py_R\results\VNE_mels"
os.makedirs(DPY_MELS_DIR, exist_ok=True)
```

```python
# MELS PARAMETERS
n_fft = 2048
hop_length = 32
n_mels = 128
fmin = 0
fmax = 11025
IMG_SIZE = 224
```

```python
# AUDIO SEGMENTS

p1_audio_segm = [f for f in os.listdir(P1_ORG_AUDIO_DIR) if f.lower().endswith(".wav")]
p2_audio_segm = [f for f in os.listdir(P2_ORG_AUDIO_DIR) if f.lower().endswith(".wav")]
print("Audio segments found in P1:", len(p1_audio_segm))
print("-" * 60)
print("Audio segments found in P2:", len(p2_audio_segm))
print("-" * 60)
```

# 2. Get Mels

```python
# RUN MELS GENERATION: P1

for idx, audio_name in enumerate(p1_audio_segm):

    print(f"[{idx+1}/{len(p1_audio_segm)}] Processing:", audio_name)

    input_audio = os.path.join(P1_ORG_AUDIO_DIR, audio_name)
    base_name = os.path.splitext(audio_name)[0]
    output_mels = os.path.join(DPY_MELS_DIR, base_name + ".png")

    try:
        # load audio
        y, sr = librosa.load(input_audio, sr=None, mono=True)

        # Mel spec
        S = librosa.feature.melspectrogram(
            y=y,
            sr=sr,
            n_fft=n_fft,
            hop_length=hop_length,
            n_mels=n_mels,
            fmin=fmin,
            fmax=fmax
        )

        # Convert to dB
        S_db = librosa.power_to_db(S, ref=np.max)

        # Normalization
        S_min, S_max = S_db.min(), S_db.max()
        if S_max - S_min < 1e-6:
            S_norm = np.zeros_like(S_db)
        else:
            S_norm = (S_db - S_min) / (S_max - S_min)

        # low and high frecuency position
        S_norm = np.flipud(S_norm)

        # color
        cmap = cm.get_cmap("inferno")
        img_array = (cmap(S_norm)[:, :, :3] * 255).astype(np.uint8)

        # image aspect: square
        img = Image.fromarray(img_array)
        img = img.resize((IMG_SIZE, IMG_SIZE), Image.BILINEAR)

        # saving
        img.save(output_mels)

        print("✅ Saved:", output_mels)
        print("-" * 40)

        del S, S_db, S_norm, img_array, img, y

    except Exception as e:
        print("❌ File error:", audio_name)
        print(e)
        print("-" * 40)

print("✅ Processing completed.")
```

```python
# RUN MELS GENERATION: P2

for idx, audio_name in enumerate(p2_audio_segm):

    print(f"[{idx+1}/{len(p2_audio_segm)}] Processing:", audio_name)

    input_audio = os.path.join(P2_ORG_AUDIO_DIR, audio_name)
    base_name = os.path.splitext(audio_name)[0]
    output_mels = os.path.join(DPY_MELS_DIR, base_name + ".png")

    try:
        # load audio
        y, sr = librosa.load(input_audio, sr=None, mono=True)

        # Mel spec
        S = librosa.feature.melspectrogram(
            y=y,
            sr=sr,
            n_fft=n_fft,
            hop_length=hop_length,
            n_mels=n_mels,
            fmin=fmin,
            fmax=fmax
        )

        # Convert to dB
        S_db = librosa.power_to_db(S, ref=np.max)

        # Normalization
        S_min, S_max = S_db.min(), S_db.max()
        if S_max - S_min < 1e-6:
            S_norm = np.zeros_like(S_db)
        else:
            S_norm = (S_db - S_min) / (S_max - S_min)

        # low and high frecuency position
        S_norm = np.flipud(S_norm)

        # color
        cmap = cm.get_cmap("inferno")
        img_array = (cmap(S_norm)[:, :, :3] * 255).astype(np.uint8)

        # image aspect: square
        img = Image.fromarray(img_array)
        img = img.resize((IMG_SIZE, IMG_SIZE), Image.BILINEAR)

        # saving
        img.save(output_mels)

        print("✅ Saved:", output_mels)
        print("-" * 40)

        del S, S_db, S_norm, img_array, img, y

    except Exception as e:
        print("❌ File error:", audio_name)
        print(e)
        print("-" * 40)

print("✅ Processing completed.")
```
